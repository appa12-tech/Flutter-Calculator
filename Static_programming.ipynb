```python
# Import the normal distribution function from the scipy.stats library
from scipy.stats import norm

# Import Matplotlib library for plotting
import matplotlib.pyplot as plt

# Generate an array of values from -3 to 3 with a step size of 0.001
x = np.arange(-3, 3, 0.001)

# Plot the Probability Density Function (PDF)
plt.plot(x, norm.pdf(x))

# Display the histogram
plt.show()
```

```python
# Import the exponential distribution function from the scipy.stats library
from scipy.stats import expon
# Import Matplotlib library for creating plots
import matplotlib.pyplot as plt

# Generate a range of x-values from 0 to 10 with small steps (0.001)
x = np.arange(0, 10, 0.001)

# Plot the Probability Density Function (PDF) of the exponential distribution
plt.plot(x, expon.pdf(x))

# Show the plot
plt.show()
```

```python
# Import the binomial distribution function from the scipy.stats library
from scipy.stats import binom

# Import Matplotlib library for plotting
import matplotlib.pyplot as plt

# Define the parameters for the binomial distribution
n, p = 10, 0.5      # n= number of trials, p= probability of success

# Generate a range of x-values from 0 to 10 with small steps (0.001) for plotting
x = np.arange(0, 10, 0.001)

# Plot the Probability Mass Function (PMF) of the binomial distribution
plt.plot(x, binom.pmf(x, n, p))

# Display the plot
plt.show()
```

```python
# Import the binomial distribution function from the scipy.stats library
from scipy.stats import poisson

# Import Matplotlib library for plotting
import matplotlib.pyplot as plt

# Define the mean (mu) of the Poisson distribution
mu = 500

# Generate a range of x-values from 400 to 600 with small steps (0.5) for plotting
x = np.arange(400, 600, 0.5)

# Plot the Probability Mass Function (PMF) of the Poisson distribution
# poisson.pmf(x, mu) calculates the PMF for the given x values
plt.plot(x, poisson.pmf(x, mu))

# Display the plot
plt.show()
```

```python
# What's the equivalent of a probability distribution function when using discrete instead of continuous data?
from collections import Counter

data = [1, 2, 3, 4, 5, 6, 1, 2, 3, 1]
counts = Counter(data)
pmf = {k: v/len(data) for k, v in counts.items()}
print(pmf)
```

```python
import numpy as np

# Generate synthetic data since pageSpeeds and purchaseAmounts are not defined
np.random.seed(42)
pageSpeeds = np.random.normal(3.0, 1.0, 1000)
purchaseAmounts = np.random.normal(50.0, 10.0, 1000) / (pageSpeeds + 0.1)

# Calculate covariance using NumPy
covariance = np.cov(pageSpeeds, purchaseAmounts)

# Display covariance matrix
print("Covariance Matrix:")
print(covariance)

# Display covariance between the two variables
print("Covariance:", covariance[0, 1])
```
